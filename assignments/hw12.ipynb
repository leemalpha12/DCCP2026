```python
import numpy as np
import pandas as pd
works = pd.DataFrame({
'작가': ['김유정', '김유정', '김유정',
'현진건', '현진건',
'이상', '이상',
'나도향', '나도향',
'채만식', '이효석'],
'작품': ['봄봄', '동백꽃', '만무방',
'운수 좋은 날', 'B사감과 러브레터',
'날개', '권태',
'벙어리 삼룡이', '물레방아',
'레디메이드 인생', '메밀꽃 필 무렵'],
'연도': [1935, 1936, 1934,
1924, 1925,
1936, 1937,
1925, 1925,
1934, np.nan], # 이효석 '메밀꽃 필 무렵' 발표연도가 누락됨
'글자수': [12500, 9800, 13600,
11200, 7400,
18300, 14100,
16800, 14500,
22000, 11800],
})
authors = pd.DataFrame({
'작가': ['김유정', '현진건', '이상', '나도향', '채만식', '이효석', '염상섭'],
'생년': [1908, 1900, 1910, 1902, 1902, 1907, 1897],
'몰년': [1937, 1943, 1937, 1926, 1950, 1942, 1963],
})
```

```python
print(works.shape)
print(works.dtypes)
print(works.describe())
```

연도 열은 결측치(NaN)가 포함되어 있어 float64로 변환되었고, 결측치가 없는 글자수는 int64이다.

```python
print(works.isna().sum())
```

```python
print(works.isna().sum())
```

```python
works2 = works.copy()

works2["연도"] = (
    works2["연도"]
    .fillna(1936)
    .astype(int)
)

print(works2.dtypes)
print(works2.tail(3))
```

```python
after_1930 = works2[works2["연도"] >= 1930]
print(after_1930)
```

```python
kim_or_lee = works2[
    works2["작가"].isin(["김유정", "이상"])
]

print(kim_or_lee)
```

```python
def categorize(n: int) -> str:
    if n < 10000:
        return "짧음"
    elif n <= 15000:
        return "보통"
    else:
        return "긴"

works2["분량"] = works2["글자수"].apply(categorize)

print(
    works2[["작품", "글자수", "분량"]]
)
```

```python
sorted_df = works2.sort_values(
    by=["연도", "글자수"],
    ascending=[True, False]
)

print(sorted_df)
```

```python
result = (
    works2.groupby("작가")["글자수"]
    .agg(["mean", "count", "sum"])
)

print(result)
```

```python
print(
    works2["분량"].value_counts()
)
```

```python
result = (
    works2.groupby(
        ["작가", "분량"]
    ).size()
)

print(result)
```

가장 많은 작품 수를 가진 조합은 (김유정, 보통, 2편): 김유정의 작품이 주로 중간 분량

```python
full = pd.merge(
    works2,
    authors,
    on="작가",
    how="left"
)

print(full)
print(full.shape)
```

확인질문: 포함되지 X. how="left"는 왼쪽 DataFrame(works2)의 행만 유지하므로 authors에만 존재하는 '염상섭'은 결과에 나타나지 X.

```python
full["집필 당시 나이"] = (
    full["연도"] - full["생년"]
)

print(
    full[
        ["작가", "작품", "연도", "생년", "집필 당시 나이"]
    ]
)
```

```python
avg_age = (
    full.groupby("작가")["집필 당시 나이"]
    .mean()
    .sort_values()
)

print(avg_age)
```

가장 어린 나이에 작품을 발표한 작가는 '나도향'으로 평균 23세에 집필함. 가장 늦은 나이에 작품을 발표한 작가는 '채만식'으로 평균 32세에 작품을 발표함.
